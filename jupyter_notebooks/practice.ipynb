```python
%%sql
SELECT 
    ROUND(
        COUNT(DISTINCT a1.player_id) * 1.0 / COUNT(DISTINCT a.player_id),
        2
    ) AS fraction
FROM 
    Activity a
LEFT JOIN 
    (SELECT player_id, MIN(event_date) AS first_login
     FROM Activity
     GROUP BY player_id) fl 
    ON a.player_id = fl.player_id
LEFT JOIN 
    Activity a1 
    ON a1.player_id = a.player_id 
   AND a1.event_date = DATE(fl.first_login + INTERVAL 1 DAY);
```

```python
%%sql
SELECT ROUND(
    SUM(CASE WHEN a1.event_date IS NOT NULL THEN 1 ELSE 0 END) * 1.0 / COUNT(DISTINCT fl.player_id),
    2) AS fraction
FROM (
    SELECT player_id, MIN(event_date) AS first_login
    FROM Activity
    GROUP BY player_id
) fl
LEFT JOIN Activity a1 
    ON a1.player_id = fl.player_id 
   AND a1.event_date = DATE(fl.first_login+ INTERVAL 1 DAY);
```
