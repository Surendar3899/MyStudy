CROSS JOIN Explanation

A CROSS JOIN is used in SQL to return the Cartesian product of two tables. In other words, it will combine each row from the first table with all the rows of the second table. The result set will contain every possible combination of rows from both tables.

For example, if:

Table 1 has 3 rows, and

Table 2 has 4 rows,

The result of a CROSS JOIN between these two tables will have 3 * 4 = 12 rows.

How It Works:

Each row from the first table is matched with every row from the second table.

If the first table has n rows, and the second table has m rows, the result of the CROSS JOIN will have n * m rows.

The result set contains columns from both tables.

Important Notes:

CROSS JOIN does not require a ON clause or a WHERE clause, since it simply returns all combinations.

It can be very resource-intensive if used carelessly, especially when working with large tables, as the result can grow very large.

Sample Scenario

Let’s use two simple tables as an example:

Table 1: students (holds a list of students)

Table 2: courses (holds a list of courses)

```python
%%sql
CREATE TABLE students (
    student_id INT PRIMARY KEY,
    student_name VARCHAR(50)
);

```

```python
%%sql
CREATE TABLE courses (
    course_id INT PRIMARY KEY,
    course_name VARCHAR(50)
);
```

```python
%%sql
-- Inserting data into the students table
INSERT INTO students (student_id, student_name) 
VALUES 
(1, 'Alice'),
(2, 'Bob'),
(3, 'Charlie');

-- Inserting data into the courses table
INSERT INTO courses (course_id, course_name) 
VALUES 
(101, 'Math'),
(102, 'Science'),
(103, 'History');
```

```python
%%sql
SELECT students.student_name, courses.course_name
FROM students
CROSS JOIN courses;
```

```python
%%sql
SELECT *
FROM students
CROSS JOIN courses;
```
