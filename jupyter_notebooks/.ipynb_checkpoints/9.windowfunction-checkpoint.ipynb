```python
%%sql
CREATE TABLE sales (
    TransactionID INT,
    Store VARCHAR(50),
    SalesAmount DECIMAL(10, 2)
);

INSERT INTO sales (TransactionID, Store, SalesAmount)
VALUES
    (1, 'A', 100.00),
    (2, 'A', 200.00),
    (3, 'A', 150.00),
    (4, 'B', 250.00),
    (5, 'B', 300.00);
```

```python
%%sql
select * from sales;
```

```python
%%sql
select *,sum(SalesAmount) over() as totalsales from sales;
```

```python
%%sql
select *,sum(SalesAmount) over(partition by store) as totalsales from sales;
```

```python
%%sql
select *,sum(SalesAmount) over(partition by store order by TransactionID DESC) as totalsales from sales;
```

🔹 1. Aggregate Functions (as Window Functions)

SUM()

AVG()

MIN()

MAX()

COUNT()

🔹 2. Ranking Functions

ROW_NUMBER()

RANK()

DENSE_RANK()

PERCENT_RANK()

NTILE(n)

🔹 3. Value Functions

LAG(expr, offset, default) → previous row value

LEAD(expr, offset, default) → next row value

FIRST_VALUE(expr) → first value in the window

LAST_VALUE(expr) → last value in the window

NTH_VALUE()

```python
%%sql
select * from sales;
```

```python
%%sql
select TransactionID,Store,SalesAmount,ROW_NUMBER() over(partition by store order by SalesAmount desc) from sales;
```

```python
%%sql
drop table if exists Employees;
CREATE TABLE Employees (
    EmployeeID INT,
    Name VARCHAR(100),
    Department VARCHAR(50),
    HireDate DATE
);


INSERT INTO Employees (EmployeeID, Name, Department, HireDate)
VALUES
    (1, 'Alice', 'HR', '2020-05-01'),
    (1, 'Alice', 'HR', '2022-06-15'),
    (2, 'Bob', 'IT', '2021-07-10'),
    (3, 'Charlie', 'Finance', '2020-09-30'),
    (3, 'Charlie', 'Finance', '2022-05-22');
```

```python
%%sql
select * from Employees;
```

```python
%%sql
WITH EmployeeBank as(select EmployeeID,Name,Department,HireDate,ROW_NUMBER() OVER(PARTITION BY EmployeeID ORDER BY HireDate) as rownum from Employees)

select * from EmployeeBank;
```

```python
%%sql
select EmployeeID from EmployeeBank;
```

```python
%%sql
WITH EmployeeBank AS (
    SELECT 
        EmployeeID,
        Name,
        Department,
        HireDate,
        ROW_NUMBER() OVER (PARTITION BY EmployeeID ORDER BY HireDate) AS rownum
    FROM Employees
)
SELECT *
FROM EmployeeBank where rownum = 1;
```

```python
%%sql
drop table if exists students;
CREATE TABLE Students (
    StudentID INT,
    StudentName VARCHAR(100),
    ExamScore INT
);

INSERT INTO Students (StudentID, StudentName, ExamScore)
VALUES
    (1, 'Alice', 95),
    (2, 'Bob', 90),
    (3, 'Charlie', 95),
    (4, 'David', 85),
    (5, 'Eva', 90);
```

```python
%%sql
select *,rank() over(order by ExamScore desc) as score_rank from Students;
```

```python
%%sql
select *,dense_rank() over(order by ExamScore desc) as score_rank from Students;
```

```python
%%sql
drop  table if exists EmployeeSales;
CREATE TABLE EmployeeSales (
    EmployeeID INT,
    EmployeeName VARCHAR(100),
    SalesAmount DECIMAL(10, 2)
);


INSERT INTO EmployeeSales (EmployeeID, EmployeeName, SalesAmount) VALUES
(1, 'Alice', 10000),
(2, 'Bob', 8500),
(3, 'Charlie', 7500),
(4, 'David', 6000),
(5, 'Eva', 11000),
(6, 'Frank', 4500),
(7, 'Grace', 3000),
(8, 'Hank', 4000),
(9, 'Ivy', 8000),
(10, 'Jack', 9500);
```

```python
%%sql
select EmployeeID,EmployeeName,SalesAmount,NTILE(4) over(order by SalesAmount desc) as groupwise from EmployeeSales;
```

```python
%%sql
drop table if exists EmployeeSalaries;
CREATE TABLE EmployeeSalaries (
    EmployeeID INT,
    EmployeeName VARCHAR(100),
    Salary DECIMAL(10, 2),
    Year INT
);
INSERT INTO EmployeeSalaries (EmployeeID, EmployeeName, Salary, Year) VALUES
(1, 'Alice', 5000, 2023),
(1, 'Alice', 5500, 2024),
(2, 'Bob', 4500, 2023),
(2, 'Bob', 4800, 2024),
(3, 'Charlie', 4000, 2023),
(3, 'Charlie', 4200, 2024),
(4, 'David', 4600, 2023),
(4, 'David', 4700, 2024),
(5, 'Eva', 5200, 2023),
(5, 'Eva', 5000, 2022),
(5, 'Eva', 5400, 2024);
```

```python
%%sql
select *,
lag(salary) over(partition by EmployeeID order by year) as previous_salary,
salary - lag(salary) over(partition by EmployeeID order by year) as difference
from EmployeeSalaries;
```

```python
%%sql
drop table if exists SalesData;
CREATE TABLE SalesData (
    SaleID INT,
    EmployeeName VARCHAR(100),
    SaleAmount DECIMAL(10, 2),
    SaleDate DATE
);
INSERT INTO SalesData (SaleID, EmployeeName, SaleAmount, SaleDate) VALUES
(1, 'Alice', 5000, '2025-01-01'),
(2, 'Bob', 3000, '2025-01-02'),
(3, 'Charlie', 4000, '2025-01-03'),
(4, 'David', 4500, '2025-01-04'),
(5, 'Eva', 5500, '2025-01-05'),
(4, 'David', 5500, '2025-01-05'),
(5, 'Eva', 6500, '2025-01-06'),
(5, 'Eva', 3000, '2025-01-07');
```

```python
%%sql
select *,
lead(SaleAmount) over(order by SaleDate) as nextsaleamount
from SalesData;
```

```python
%%sql
select *,
lead(SaleAmount) over(order by SaleDate) as nextDatesaleamount,
lead(SaleAmount) over(order by SaleDate) - SaleAmount difference
from SalesData;
```
