```python
%%sql
drop table if exists employees;
```

```python
%%sql
create table employees(employee_id INT PRIMARY KEY,first_name VARCHAR(50) NOT NULL,last_name VARCHAR(50) NOT NULL,hire_date DATE NOT NULL,salary DECIMAL(10,3))
```

```python
%%sql
insert into employees(employee_id,first_name,last_name,hire_date,salary) values (1,"surendar","nadarajan",'2025-08-10',50000),(2,"susi","nadarajan",'2025-08-10',70000),(3,"aravind","jayaraman",'2025-08-10',60000),(4,"dinesh","ravi",'2024-08-10',30000);
```

```python
%%sql
select * from employees;
```

```python
%%sql
#CTAS (create Table AS)
create table high_paid_employees as(select employee_id,first_name,last_name,salary from employees where salary > 50000)
```

```python
%%sql
select * from high_paid_employees;
```

```python
%%sql
create temporary table temp_high_paid_employees(employee_id int,firstname varchar(50))
```

```python
%%sql
insert into temp_high_paid_employees select employee_id,first_name from employees where salary > 50000;
```

```python
%%sql
select * from temp_high_paid_employees;
```

```python
%%sql
#CTE(common Table Expression)
WITH high_salary_cte as(SELECT employee_id,first_name,last_name,salary FROM employees where salary > 50000)

#create table high_salary_employees_via_cte select * from high_salary_cte;
```

```python
%%sql
SELECT VERSION();
```

```python
%%sql
WITH high_salary_cte AS (
    SELECT employee_id, first_name, last_name, salary 
    FROM employees 
    WHERE salary > 50000
)
CREATE TABLE high_salary_employees_via_cte 
SELECT * FROM high_salary_cte;
```
