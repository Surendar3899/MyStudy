```python
%%sql
SELECT m.id AS manager_id,
       m.name AS manager_name,
       AVG(e.salary) AS avg_team_salary
FROM employees e
JOIN employees m 
    ON e.manager_id = m.id
WHERE e.designation != 'Manager'  -- exclude managers from team members
GROUP BY m.id, m.name
ORDER BY avg_team_salary DESC
LIMIT 1;
```

```python
%%sql
SELECT m.id AS manager_id,
       m.name AS manager_name,
       MAX(e.salary) AS avg_team_salary
FROM employees e
JOIN employees m 
    ON e.manager_id = m.id
WHERE e.designation != 'Manager'  -- exclude managers from team members
GROUP BY m.id, m.name LIMIT 1;
```

```python
%%sql
select m.name,count(e.id) from employees m join employees e on m.id = e.manager_id where e.designation != 'Manager' GROUP BY m.id;
```

```python
%%sql
SELECT m.name AS manager_name,
       COUNT(e.id) AS employee_count
FROM employees m
JOIN employees e 
    ON m.id = e.manager_id
WHERE e.designation != 'Manager'
GROUP BY m.id;
```

```python
%%sql
#Find employees who joined before 6 months
SELECT * 
FROM employees 
WHERE DATEDIFF(CURDATE(), joining_date) > 180
order by joining_date desc;
```

```python
%%sql
#Find employees who joined in the last 6 months
SELECT * 
FROM employees
WHERE joining_date >= DATE_SUB(CURDATE(), INTERVAL 6 YEAR);
```
